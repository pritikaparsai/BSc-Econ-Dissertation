```python
import pandas as pd
import numpy as np
```

```python
path = "/Users/Pritika/Downloads/Data_in_CSV/cperv12025.csv"

df = pd.read_csv(path)
```

```python
df = df.rename(columns={
    'panel': 'Panel',
    'file_id': 'File_ID',
    'sec': 'Sector',
    'st': 'State_Code',
    'visit': 'Visit',
    'sex': 'Sex',
    'age': 'Age',
    'gedu_lvl': 'Education_Level',
    'pas': 'UPSS_Status',
    'ind_pas': 'NIC_Code',
    'ocu_pas': 'NCO_Code',
    'ern_reg': 'Earnings',
    'mult': 'Weight'
})
```

```python
print("--- Age ---")
print(df['Age'].describe())

print("\n--- Sex ---")
print(df['Sex'].value_counts())

print("\n--- Sector (Rural/Urban) ---")
print(df['Sector'].value_counts())

print("\n--- State_Code ---")
print(df['State_Code'].nunique())

print("\n--- UPSS_Status  ---")
print(df['UPSS_Status'].value_counts())

print("\n--- NCO_Code  ---")
print(df['NCO_Code'].nunique())

print("\n--- Earnings ---")
print(df['Earnings'].describe())
```

```python
df['NCO_Code'] = df['NCO_Code'].astype(str).str.strip().str.zfill(3)
df['NIC_Code'] = df['NIC_Code'].astype(str).str.strip()

df['Earnings'] = pd.to_numeric(df['Earnings'], errors='coerce')
df['Weight'] = pd.to_numeric(df['Weight'], errors='coerce')

df.dtypes
```

```python
df.to_csv('../outputs/raw_loaded.csv', index=False)
print("Saved. Shape:", df.shape)
```
